برای مقداردهی تابع با مشخصاتی که دادی، ابتدا باید پارامتر زمان را بر اساس ویژگی‌های فیزیکی پاندول محاسبه کنیم تا نوسان در مقصد حذف شود.

۱. محاسبه پارامترهای زمانی
فرکانس طبیعی و دوره تناوب پاندول شما به این ترتیب است:

   فرکانس طبیعی ($\omega_n$):*
    $$\omega_n = \sqrt{\frac{g}{L}} = \sqrt{\frac{9.80665}{0.3}} \approx 5.717 \text{ rad/s}$$
   دوره تناوب ($T$):*
    $$T = \frac{2\pi}{\omega_n} \approx 1.099 \text{ s}$$

برای اینکه نوسان در انتهای مسیر صفر باشد، زمان کل حرکت ($t_f$) باید ضریبی از دوره تناوب باشد. ساده‌ترین حالت این است که زمان کل را برابر با یک دوره تناوب کامل در نظر بگیریم:
$t_f = 1.099$ ثانیه

۲. جایگذاری در تابع حرکت $x(t)$
با استفاده از مدل S-curve که نرم‌ترین حرکت را ایجاد می‌کند و با فرض $x_d = 0.1$ متر، تابع جابجایی ارابه بر حسب زمان به صورت زیر به دست می‌آید:

$$x(t) = \frac{0.1}{1.099} \left[ t - \frac{1.099}{2\pi} \sin\left(\frac{2\pi t}{1.099}\right) \right]$$

اگر محاسبات عددی را ساده کنیم، تابع نهایی شما در بازه زمانی $0 \le t \le 1.099$ به این شکل خواهد بود:

$$x(t) \approx 0.091 \cdot t - 0.0159 \cdot \sin(5.717 \cdot t)$$

۳. راهنمای استفاده
   بازه زمانی:* این تابع فقط برای بازه زمانی صفر تا ۱.۰۹۹ ثانیه معتبر است. بعد از این زمان، $x(t)$ باید ثابت و برابر با $0.1$ باقی بماند.
   وضعیت پاندول:* با این پروفایل، پاندول در طول مسیر نوسان کوتاهی می‌کند اما دقیقاً در ثانیه ۱.۰۹۹، وقتی ارابه در مکان ۰.۱ متر متوقف می‌شود، زاویه تتا ($\theta$) صفر شده و هیچ حرکتی نخواهد داشت.
   نکته:* اگر می‌خواهی حرکت آرام‌تر انجام شود، می‌توانی زمان کل را دو برابر دوره تناوب در نظر بگیری ($t_f = 2.198 \text{ s}$)، که در این صورت نوسانات پاندول در طول مسیر حتی کمتر هم خواهد شد.

$$x(t) = \frac{x_d}{t_f} \left[ t - \frac{t_f}{2\pi} \sin\left(\frac{2\pi t}{t_f}\right) \right]$$
